## The problem

**Pascal's triangle.** Each row starts and ends with 1. Every other number is the sum of the two numbers above it.

```
1
1 1
1 2 1
1 3 3 1
1 4 6 4 1
```

Example: in the last row, 6 = 3 + 3.

**Paths in a grid.** You walk from the top-left corner to the bottom-right corner. You can only go **right** or **down**. Let `ways[i][j]` be the number of paths to the cell in row `i` and column `j`. Then:

$$\text{ways}[i][j] = \text{ways}[i-1][j] + \text{ways}[i][j-1]$$

The first row and the first column are all 1. Here is the table for a grid with 2 rows and 2 columns of steps. The answer is the last number, 6.

```
1 1 1
1 2 3
1 3 6
```

# --------------TASK 1------------------

````python
def pascal_row(n):
    row = [1]
    for i in range(n):
        new_row = [1]
        for k in range(1, len(row)):

            new_row.append(row[k - 1] + row[k])
        new_row.append(1) # J'ajoute toujours 1 à la fin 
        row = new_row
    return row

pascal_row(5)
````

# --------------TASK 3------------------

````python

def grid_paths(rows, cols):
    # Remplissons le tableau de 1
    ways = []
    for i in range(rows + 1):
        ligne = []
        for j in range(cols + 1):
            ligne.append(1)
        ways.append(ligne)
    # 2. Remplir les cases avec la règle haut + gauche
    for i in range(1, rows + 1):
        for j in range(1, cols + 1):
            ways[i][j] = ways[i - 1][j] + ways[i][j - 1]# la formules 
    # 3. Renvoyer la case en bas à droite
    return ways[rows][cols]
````

# --------------TASK 4------------------

````python
def grid_paths_blocked(rows, cols, blocked):
# crée la liste des elements bloqués 
    blocked = set(blocked)

    # 1. Construire le tableau rempli de 0
    ways = []
    for i in range(rows + 1):
        ligne = []
        for j in range(cols + 1):
            ligne.append(0)
        ways.append(ligne)

    # 2. Remplir chaque case
    for i in range(rows + 1):
        for j in range(cols + 1):
            if (i, j) in blocked:
                ways[i][j] = 0
            elif i == 0 and j == 0:
                ways[i][j] = 1
            else:
                from_top = ways[i - 1][j] if i > 0 else 0
                from_left = ways[i][j - 1] if j > 0 else 0
                ways[i][j] = from_top + from_left
    return ways[rows][cols]
# Task 4: 4 x 4 grid with two blocked cells
print()
print("4 x 4 grid, blocked = [(1, 2), (2, 1)]:", grid_paths_blocked(4, 4, [(1, 2), (2, 1)]))
````

````python
def count_odd(row):
    """Return how many odd numbers are in a list."""
    compteur = 0
    for x in row:
        if x % 2 == 1:
            compteur = compteur + 1
    return compteur
counts = []
for n in range(16):
    nombre = count_odd(pascal_row(n))
    counts.append(nombre)

print("Rows 0 to 15:", counts)

# Guess for row 16: 16 = 10000 in binary (one 1), so 2 ** 1 = 2.
print("Row 16:", count_odd(pascal_row(16)))
````

## Extra (optional)

Print Pascal's triangle with a `*` for an odd number and a space for an even number (16 rows). What shape do you see?

````python
def print_odd_triangle(n_rows):
    # Print Pascal's triangle with '*' for odd numbers and ' ' for even numbers."""
    for n in range(n_rows):
        line = " ".join("*" if x % 2 == 1 else " " for x in pascal_row(n))
        print(" " * (n_rows - 1 - n) + line)
print_odd_triangle(16)
````

**Answer.** We see the **Sierpinski triangle**: a big triangle made of three smaller copies of itself, with an empty (even) triangle in the middle. Each copy is again made of three smaller copies, and so on.
